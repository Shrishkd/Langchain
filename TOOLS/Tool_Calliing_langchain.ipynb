# Currecy Conversion

```python
from langchain_core.tools import InjectedToolArg
from typing import Annotated
```

```python
# tool create
@tool
def get_conversion_factor(base_currency: str, target_currency: str) -> float:
  """
  This function fetches the currency conversion factor between a given base currency and a target currency
  """
  url = f'https://v6.exchangerate-api.com/v6/3c6b25078af9e146f8c2416d/pair/{base_currency}/{target_currency}'

  response = requests.get(url)

  return response.json()

@tool
def convert(base_currency_value: int, conversion_rate: Annotated[float, InjectedToolArg]) -> float:
  """
  given a currency conversion rate this function calculates the target currency value from a given base currency value
  """

  return base_currency_value * conversion_rate
```

```python
convert.args
```

```python
get_conversion_factor.invoke({'base_currency':'USD','target_currency':'INR'})
```

```python
convert.invoke({'base_currency_value':10, 'conversion_rate': 95.2096})
```

```python
#tool binding
llm = ChatGoogleGenerativeAI(model="gemini-3.6-flash")
```

```python
llm_with_tools = llm.bind_tools([get_conversion_factor, convert])
```

```python
messages = [HumanMessage('What is the conversion factor between INR and USD, and based on that can you convert 10 inr to usd')]
```

```python
messages
```

```python
ai_message = llm_with_tools.invoke(messages)
```

```python
ai_message
```

```python
messages.append(ai_message)
```

```python
ai_message.tool_calls
```

```python
import json

for tool_call in ai_message.tool_calls:
  # execute the 1st tool and get the value of conversion rate
  if tool_call['name'] == 'get_conversion_factor':
    tool_message1 = get_conversion_factor.invoke(tool_call)
    # fetch this conversion rate
    conversion_rate = json.loads(tool_message1.content)['conversion_rate']
    # append this tool message to messages list
    messages.append(tool_message1)
  # execute the 2nd tool using the conversion rate from tool 1
  if tool_call['name'] == 'convert':
    # fetch the current arg
    tool_call['args']['conversion_rate'] = conversion_rate
    tool_message2 = convert.invoke(tool_call)
    messages.append(tool_message2)

```

```python
messages
```

```python
tool_result = llm_with_tools.invoke(messages)
```

### Corrected Currency Conversion Workflow

```python
# Start a new conversation for clarity
messages_corrected = [HumanMessage('What is the conversion factor between INR and USD, and based on that can you convert 10 inr to usd')]
display(messages_corrected)
```

```python
# First LLM invocation: The LLM should identify the need for 'get_conversion_factor'
first_ai_message = llm_with_tools.invoke(messages_corrected)
messages_corrected.append(first_ai_message)
display(messages_corrected)
```

```python
# Execute the 'get_conversion_factor' tool and append its result
# We expect only one tool call here from the first_ai_message
if first_ai_message.tool_calls and first_ai_message.tool_calls[0]['name'] == 'get_conversion_factor':
    tool_call_get_conversion = first_ai_message.tool_calls[0]
    tool_message_conversion_factor = get_conversion_factor.invoke(tool_call_get_conversion)
    messages_corrected.append(tool_message_conversion_factor)

    # Extract conversion rate for potential direct use, though LLM should handle it
    conversion_rate = json.loads(tool_message_conversion_factor.content)['conversion_rate']
    print(f"Fetched conversion rate: {conversion_rate}")
else:
    print("LLM did not propose 'get_conversion_factor' as expected.")
display(messages_corrected)
```

```python
# Second LLM invocation: Now the LLM has the conversion rate in its context and should propose 'convert'
second_ai_message = llm_with_tools.invoke(messages_corrected)
messages_corrected.append(second_ai_message)
display(messages_corrected)
```

```python
# Execute the 'convert' tool and append its result
# We expect only one tool call here from the second_ai_message
if second_ai_message.tool_calls and second_ai_message.tool_calls[0]['name'] == 'convert':
    tool_call_convert = second_ai_message.tool_calls[0]
    # Manually inject the conversion_rate as the LLM did not provide it in the tool call args
    tool_call_convert['args']['conversion_rate'] = conversion_rate
    tool_message_converted_value = convert.invoke(tool_call_convert)
    messages_corrected.append(tool_message_converted_value)
else:
    print("LLM did not propose 'convert' as expected.")
display(messages_corrected)
```

```python
# Final LLM invocation: Get the natural language response using the full conversation history
final_response = llm_with_tools.invoke(messages_corrected).content
print(final_response)
```
